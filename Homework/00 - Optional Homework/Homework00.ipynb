```python
#importing data
dataset_path = './interactive_data.csv'
"""
. is for working directory 
"""

#reading dataset as pandas dataframe
df = pd.read_csv(dataset_path, index_col=0) 
"""
index_col = 0 tells pandas to use the first column of the CSV (column position 0, since Python indexing starts at 0) 
as the row labels (index) of the resulting DataFrame, instead of treating it as a regular data column.
"""
df.head(5)
```

the second row is the total white population (any white people since it's the only criteria)

### Nearly *two-thirds* of gun deaths are *suicides*.

I need to get the total number of gun deaths and the total number of suicide deaths.

```python
#using loc function with boolean conditions to select the needed value
total_gundeaths = df.loc[
                        (df['Intent'] == 'None selected') & 
                        (df['Gender'] == 'None selected') & 
                        (df['Age'] == 'None selected') & 
                        (df['Race'] == 'None selected')
                    , 'Deaths'].iloc[0]
total_suicidedeaths = df.loc[
                            (df['Intent'] == 'Suicide') & 
                            (df['Gender'] == 'None selected') & 
                            (df['Age'] == 'None selected') & 
                            (df['Race'] == 'None selected')
                        , 'Deaths'].iloc[0]
print("The total number of gundeaths is", total_gundeaths,  "and the total number of suicide gun deaths is", total_suicidedeaths)
print("The ratio of suicide deaths is", total_suicidedeaths/total_gundeaths*100, "%")


#a little more complicated but same thing
#total_gundeaths2 = df.loc[(df[['Intent', 'Gender', 'Age', 'Race']] == 'None selected').all(axis=1), 'Deaths'].iloc[0]
#total_suicidedeaths2 = df.loc[(df[['Gender', 'Age', 'Race']] == 'None selected').all(axis=1) & (df['Intent'] == 'Suicide'), 'Deaths'].iloc[0]
#print("The total number of gundeaths is %(tgd)d and the total number of suicide gun deaths is %(tsd)d" % {"tgd": total_gundeaths2, "tsd": total_suicidedeaths2})
```

### More than *85 percent* of suicide victims are *male*.

I need to get the total number of male suicide victims.

```python
totat_male_suicide_gundeaths = df.loc[
                                    (df['Intent'] == 'Suicide') & 
                                    (df['Gender'] == 'Male') & 
                                    (df['Age'] == 'None selected') & 
                                    (df['Race'] == 'None selected')
                                , 'Deaths'].iloc[0]

print(totat_male_suicide_gundeaths/total_suicidedeaths*100, "percent of suicide victims are male")
```

### Around *a third* of all gun deaths are *homicides*.

I need to get the total number of homicides.

```python
totat_homicide_gundeaths = df.loc[
                                (df['Intent'] == 'Homicide') & 
                                (df['Gender'] == 'None selected') & 
                                (df['Age'] == 'None selected') & 
                                (df['Race'] == 'None selected')
                            , 'Deaths'].iloc[0]

print(totat_homicide_gundeaths/total_gundeaths*100, "percent of suicide victims are male")
```

### Around *two-thirds* of homicide victims who are *males* in the *age-group of 15--34* are *black*.

I need to get the homicide victims who are black males between 15 and 34.

```python
homicide_male_1534_black = df.loc[
                                (df['Intent'] == 'Homicide') & 
                                (df['Gender'] == 'Male') & 
                                (df['Age'] == '15 - 34') & 
                                (df['Race'] == 'Black')
                            , 'Deaths'].iloc[0]

print(homicide_male_1534_black/totat_homicide_gundeaths*100, "percent of homicide victims are ...")
```

### *Women* constitue only *15 percent* of the total *homicide* victims.

I need to get homicide victims who are female.

```python
homicide_female = df.loc[
                        (df['Intent'] == 'Homicide') & 
                        (df['Gender'] == 'Female') & 
                        (df['Age'] == 'None selected') & 
                        (df['Race'] == 'None selected')
                , 'Deaths'].iloc[0]

print(homicide_female/totat_homicide_gundeaths*100, "percent of homicide victims are female")
```
