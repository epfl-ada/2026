## Data from World Happiness Report

The World Happiness Report is an annual publication of the United Nations Sustainable Development Solutions Network. It contains articles, and rankings of national happiness based on respondent ratings of their own lives, which the report also correlates with various life factors.

In this notebook we will explore the happiness of different countries and the features associated.
The datasets that we will use are available in *Data*: **happiness2020.pkl** and **countries_info.csv**.

Although the features are self-explanatory, here a summary: 

**happiness2020.pkl**
* country: *Name of the country*
* happiness_score: *Happiness score*
* social_support: *Social support (mitigation the effects of inequality)*
* healthy_life_expectancy: *Healthy Life Expectancy*
* freedom_of_choices: *Freedom to make life choices*
* generosity: *Generosity (charity, volunteers)*
* perception_of_corruption: *Corruption Perception*
* world_region: *Area of the world of the country*

**countries_info.csv**
* country_name: *Name of the country*
* area: *Area in sq mi*
* population: *Number of people*
* literacy: *Literacy percentage*

```python
import pandas as pd
%matplotlib inline
# IPython/Jupyter "magic command" that tells Jupyter to render any matplotlib plots directly inside the notebook

DATA_FOLDER = 'Data/'

HAPPINESS_DATASET = DATA_FOLDER+"happiness2020.csv"
COUNTRIES_DATASET = DATA_FOLDER+"countries_info.csv"
#fancy way of just doing a regular variable assignment (variable = variable path as a string)
```

## Task 1: Load the data

Load the 2 datasets in Pandas dataframes (called *happiness* and *countries*), and show the first rows.

```python
happiness_df = pd.read_csv(HAPPINESS_DATASET)

countries_df = pd.read_csv(COUNTRIES_DATASET)

"""
print(happiness_df.head(2))
print("\n")
print(countries_df.head())
"""
#Jupyter automatically displays the last expression in a cell nicely (with the styled HTML table), but only the last one. To show two DataFrames with real visual spacing and full pandas styling (not just plain text), use display()
from IPython.display import display

display(happiness_df.head())
print()  # adds a blank line
display(countries_df.head())
```

## Task 2: Let's merge the data

Create a dataframe called *country_features* by merging *happiness* and *countries*. A row of this dataframe must describe all the features that we have about a country.

**Hint**: Verify that all the rows are in the final dataframe.

```python
#renaming the key column to be able to merge the two dataframes
countries_df = countries_df.rename(columns = {"country_name": "country"})

#putting all the country names in lowercase so that the key column values match
happiness_df["country"] = happiness_df["country"].str.lower()

#merging the two datasets into a new one call country features
country_features = pd.merge(happiness_df, countries_df, on="country", how="outer")
display(country_features.head())

"""
#we can do the same thing without changing the initial datasets:
country_features = pd.merge(
    happiness_df.assign(country=happiness_df["country"].str.lower()), 
    countries_df,
    on="country"
    )
"""

```

## Task 3: Where are people happier?

Print the top 10 countries based on their happiness score (higher is better).

```python
#sorting the dataset by descending happiness_score value
countries_happiness_score_sorted = country_features.sort_values(by="happiness_score", ascending=0)

display(countries_happiness_score_sorted.head(10))
```

We are interested to know in what world region people are happier. 

Create and print a dataframe with the (1) average happiness score and (2) the number of contries for each world region.
Sort the result to show the happiness ranking.

```python
#converting the literacy column from string to float (commas were being used as decimal points)
country_features["literacy"] = (
    country_features["literacy"]
    .str.replace(",", ".")
    .astype(float)
)
"""
Literacy values were stored as strings because they used a comma instead of a period. 
str.replace(",", ".") swaps the comma for a period in every value
astype(float) converts the column from string to a proper numeric type
"""

#creating a new region data set with the mean of all the values from the same region
world_region_features = (
    country_features
    .drop(columns="country")
    .groupby("world_region")
    .mean()
    .reset_index()
)
"""
First, drop() removes the country column.

Secondly, groupby() partitions the DataFrame's rows into separate buckets (one bucket per unique region value)
and creates a special object called a DataFrameGroupBy.

Then, mean() gets applied independently to each bucket.

Finally, reset_index() restores world_region as a normal column 
because groupby() turns the grouping column (world_region) into the index rather than keeping it as a regular column.
"""
#adding the "number of countries per region" column
world_region_features["num_countries"] = (
    country_features
    .groupby("world_region")["country"]
    .count()
    .values
)
display(world_region_features)
"""
count() returns an indexed pandas Series
=> values() strips away the Series' index and gives you just the raw numbers in order
"""

#creating a dataframe with average happiness score and number of countries per world region
happiness_by_region = (
    country_features
    .groupby("world_region")
    .agg(
        happiness_score=("happiness_score", "mean"),
        num_countries=("country", "count")
    )
    .reset_index()
    .sort_values(by="happiness_score", ascending=False)
)
"""
agg() applies a different function to each column, specified as name=(column, function):
- happiness_score=("happiness_score", "mean") computes the average happiness_score per region
- num_countries=("country", "count") counts how many (non-null) country entries are in each
  region — since every row has exactly one country, this gives the number of countries per region
"""

display(happiness_by_region)
```

The first region has only a few countries! What are them and what is their score?

```python
#finding the countries that share the "North America and ANZ" region and getting their score
north_america_and_anz_countries_happiness = (
    country_features
    .loc[country_features["world_region"] == "North America and ANZ", 
        ["country", "happiness_score"]]
)
"""
loc[row selection, column selection]
row selection: boolean condition filter
column selection: list of selected columns
"""

display(north_america_and_anz_countries_happiness)
```

## Task 4: How literate is the world?

Print the names of the countries with a level of literacy of 100%. 

For each country, print the name and the world region in the format: *{region name} - {country name} ({happiness score})*

```python
#finding the countries that share 100% literacy and getting their region, country and happiness score
full_literacy_countries = (
    country_features
    .loc[country_features["literacy"] == 100, 
         ["world_region","country", "happiness_score"]]
)

#printing each country in the format: region - country (happiness score)
for index, row in full_literacy_countries.iterrows():
    print(f"{row["world_region"]} - {row["country"]} ({row["happiness_score"]})")

"""
iterrows() is a DataFrame method that allows looping over rows one at a time
f string formatting embeds variable values directly inside {} placeholders within the text
"""
display(full_literacy_countries)
```

What is the global average?

```python
print(f"The literacy global average is {country_features.mean(axis=0, numeric_only=1).loc["literacy"]} %.")
```

Calculate the proportion of countries with a literacy level below 50%. Print the value in percentage, formatted with 2 decimals.

```python
#counting number of countries with literacy level below 50%
a = len(country_features.loc[country_features["literacy"] < 50, ["literacy"]])

#counting total number of countries
b = len(country_features)

"""
len(df) returns the number of entries in the DataFrame's index, which equals the number of rows
len(df.columns) to get number of columns

or df.shape[0 or 1], 0 for index and 1 for columns
"""

#calculating proportion of countries with literacy level below 50%
c = a/b*100

print(f"The proportion of countries with a literacy level below 50% is {c:.2f}%")
```

Print the raw number and the percentage of world population that is illiterate.

```python
#number of illiterate people per country = population - population * literacy%/100
illiterate_per_country = country_features["population"] - country_features["population"] * country_features["literacy"] / 100

#total illiterate people worldwide
raw_number = illiterate_per_country.sum()

#total world population
total_population = country_features["population"].sum()

#percentage of world population that is illiterate
percentage = raw_number / total_population * 100

print(f"The total number of illiterate people in the world is {raw_number:.0f} and the percentage is {percentage:.2f}%.")
```

## Task 5: Population density

Add to the dataframe a new field called *population_density* computed by dividing *population* by *area*.

```python
# creating new field
population_density = country_features["population"] / country_features["area"]

# adding it to the dataframe
country_features["population_density"] = population_density

display(country_features.head(5))
```

What is the happiness score of the 3 countries with the lowest population density?

```python
# sorting the dataframe by population density
pop_dens_sorted = country_features.sort_values(by="population_density")

# getting the happiness score of the 3 lowest
three_lowest_density_happiness = pop_dens_sorted.head(3)[["country", "population_density", "happiness_score"]]
display(three_lowest_density_happiness)
```
